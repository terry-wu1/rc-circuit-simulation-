```python
import matplotlib.pyplot as plt
import math

time=[]
voltage=[]

totaltime=float(input("Total time?"))
numinterval=int(input("Number of intervals?"))
initialcharge=float(input("Initlal charge?"))
capacitance=float(input("Capacitance?"))
resistance=float(input("Resistance?"))

initialvoltage=initialcharge/capacitance
timeperinterval=totaltime/numinterval

t=0



while t <= totaltime:
  voft=initialvoltage*math.exp(-t/(resistance*capacitance))
  time.append(t)
  voltage.append(voft)
  t=t+timeperinterval
1



plt.plot(time,voltage)
plt.xlabel("time")
plt.ylabel("voltage")
plt.title("voltage of discharging capacitor")
plt.grid(True)
plt.show()
```
